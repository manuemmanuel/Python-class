# Basic Concepts in Image Processing


1. **Analog vs. Digital Information**

Analog information is continuous (e.g., time on an analog clock, sound waves on a vinyl record, or light intensity in a photograph).

Digital information consists of discrete values (e.g., numbers on a digital clock, text characters, or digital images).

Converting an analog image into a digital format requires breaking continuous information into discrete units.

**2.Sampling and Digitization of Images**

A visual scene contains infinite color and intensity values.

To digitize an image, discrete points (pixels) are sampled on a two-dimensional grid.

Higher sampling rates (more pixels) improve image quality but have practical limits based on human perception.

**3. Pixel Density and Image Quality**

The human eye cannot distinguish objects closer than 0.1 mm apart.

A sufficient sampling rate is 10 pixels per millimeter or 250 pixels per inch (PPI).

A 3x5 inch image at 250 PPI requires 937,500 pixels (~1 megapixel) for high accuracy.

Lower resolutions can still provide acceptable image quality for most purposes.

# Image File Formats and Compression

**1. Raw Image Files**

Stores all sampled data, preserving the highest quality.

Large file size due to uncompressed data.

Useful for professional photography but inefficient for storage and sharing.

**2. Compressed Image Formats**

Developed to reduce file size for storage and faster network transmission.
Two widely used formats: JPEG and GIF.

**3. JPEG (Joint Photographic Experts Group)**

Uses lossless and lossy compression techniques.

*Lossless compression*: Stores position data instead of duplicate colors, preserving image quality.

*Lossy compression:* Approximates larger regions of colors, reducing file size while maintaining visual quality.

Suitable for photographs and detailed images.

**4. GIF (Graphics Interchange Format)**

Uses a color palette-based compression (max 256 colors).

Replaces color values with integer keys from a limited palette.

Efficient for flat-color images like cartoons, icons, and banners.

Lossy compression but optimized for specific image types.

# Properties of Images

**Image Representation in Programs**


When an image is loaded, software maps its bits into a rectangular grid of colored pixels.

The grid’s coordinates range from (0,0) in the upper-left to (width – 1, height – 1) in the lower-right.

This differs from the Cartesian coordinate system, where (0,0) is typically the center.

**Image Dimensions**

Defined by width (number of pixels horizontally) and height (number of pixels vertically).

Each pixel has a specific position (x, y) in the grid.

**Color Representation: RGB Model**

Each pixel’s color is defined by an (r, g, b) tuple, where:

    r = Red component (0-255)
    
    g = Green component (0-255)
    
    b = Blue component (0-255)

Combining these values produces millions of possible colors.

image.png

**1. Coordinate System (x, y)**

The top-left pixel is at (0,0) and the bottom-right pixel is at (width-1, height-1).

The x-axis moves rightward, and the y-axis moves downward, just as described in the SCSS.

**2. RGB Color Representation**

Each pixel is identified by its (x, y) coordinates and has a corresponding (r, g, b) color value.

    Example from the image:

    (1,1) → (255,255,0) (Yellow)
    
    (5,2) → (51,51,255) (Blue)
    
    (0,5) → (57,181,74) (Greenish)

**3. Image Manipulation Relevance**

The color and position of each pixel can be modified using image-processing techniques like color filtering, brightness adjustment, and edge detection.

Software like Photoshop, OpenCV, and PIL use this grid-based approach to process images.

**What is Edge Detection?**

Edge detection is a technique used in image processing and computer vision to identify boundaries between objects in an image. It highlights sudden intensity changes (contrast) in an image, which usually indicate object edges.

Why is Edge Detection Important?

✔ Helps in object detection & recognition

✔ Used in medical imaging (X-rays, MRIs, etc.)

✔ Used in autonomous driving (lane detection, obstacle detection)

✔ Key step in image segmentation & feature extraction

```python
# Load the image
image = cv2.imread("/content/drive/MyDrive/Photo/cars.jpg")

# Convert the image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Canny Edge Detection (threshold1 = 100, threshold2 = 200)
#100 and 200 are the two threshold values:
#threshold1 (100): Lower threshold – pixels with gradient magnitude below this are rejected.
#threshold2 (200): Upper threshold – pixels with a gradient magnitude above this are accepted as edges.
edges = cv2.Canny(gray, 100, 200)

# Display result
import matplotlib.pyplot as plt # Import the matplotlib library
plt.imshow(edges, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```

# **Key Concepts of Event-Driven Programming**

Event-driven programming is a programming paradigm in which the flow of the program is determined by events such as user actions (mouse clicks, keyboard input), sensor outputs, or messages from other programs. It is widely used in graphical user interfaces (GUIs), real-time systems, and applications that require a high level of interactivity.

Key Concepts of Event-Driven Programming


**1. Event Loop:**

The event loop runs continuously, waiting for events to occur. When an event happens, it is detected and processed. For instance, in a GUI application, the event loop listens for user actions like button clicks or keyboard input.

**2. Events:**

Events are actions or occurrences that happen during the program's execution. Examples include mouse clicks, key presses, and messages from other programs or sensors.

**3. Event Handlers (or Callback Functions):**

These are functions or methods that are executed in response to specific events. When an event occurs, the event loop calls the corresponding event handler to process the event.

**4. Widgets:**

Widgets are elements of the user interface such as windows, buttons, labels, and text input fields. They generate events when the user interacts with them.

**5. Event Binding:**

Event binding connects events to event handlers. In a GUI framework like Tkinter, you can bind events to callback functions using parameters like command in a Button widget.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import ipywidgets as widgets
from IPython.display import display, clear_output

# Function to open and display a single image
def open_image(change):
    uploaded_file = change['new']
    clear_output(wait=True)  # Clear previous outputs

    # Display widgets again
    display(upload_button, message_button)

    if uploaded_file:
        # Get the first (single) file
        file_name = list(uploaded_file.keys())[0]

        # Save uploaded file locally
        with open(file_name, "wb") as f:
            f.write(uploaded_file[file_name]["content"])  # Write file content to disk

        print(f"📂 Opened: {file_name}")
        image = Image.open(file_name)

        # Display the image
        plt.figure(figsize=(4, 4))
        plt.imshow(image)
        plt.axis("off")
        plt.show()

# Function to show a popup message
def show_message(_):
    print("📢 Hello! This is a popup message.")

# Upload Button (Allows only one file)
upload_button = widgets.FileUpload(accept="image/*", multiple=False)  # Single file upload
upload_button.observe(open_image, names='value')

# Message Button
message_button = widgets.Button(description="Show Message")
message_button.on_click(show_message)

# Display Widgets in a vertical box
display(widgets.VBox([upload_button, message_button]))

```
