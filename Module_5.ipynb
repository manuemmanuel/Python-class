**📌 What is an API?**

An API (Application Programming Interface) is a set of rules and protocols that allows different software applications to communicate with each other. It acts as a bridge that lets one application access data or functionality from another application or service, usually via HTTP requests (like GET, POST, PUT, DELETE).



This code makes a GET request to the JokeAPI (https://v2.jokeapi.dev/joke/Any) to fetch a random joke in JSON format. It then checks whether the joke is a single joke or two-part joke and formats the response accordingly before printing it.

```python
import requests

def get_joke():
    url = "https://v2.jokeapi.dev/joke/Any"
    response = requests.get(url)

    if response.status_code != 200:
        return "Failed to fetch joke."

    data = response.json()

    if data["type"] == "single":
        return data["joke"]
    else:
        return f'{data["setup"]} - {data["delivery"]}'

# Example usage
joke = get_joke()
print("Here's a joke for you:")
print(joke)
```

***📌 How API Keys Work***

**🔑 How to Use an API Key:**

Obtain the API key from the provider (e.g., Weatherstack, Google, etc.) after signing up.

Include the API key in your requests, typically as:

    A URL query parameter (e.g., ?access_key=YOUR_API_KEY).

    A Header (e.g., Authorization: Bearer YOUR_API_KEY).


The API will reject requests if the key is missing, incorrect, or expired.

```python
# Creating a 1D array
arr1 = np.array([1, 2, 3])

# Creating a 2D array
arr2 = np.array([[1, 2, 3], [4, 5, 6]])

# Creating an array with zeros
zeros = np.zeros((2, 3))

# Creating an array with ones
ones = np.ones((3, 2))

# Creating an array with a range of values
#📌 Breakdown:
#0 → Starting value (inclusive).
#10 → Stopping value (exclusive).
#2 → Step size (difference between consecutive elements).

arr3 = np.arange(0, 10, 2)
```

**📌 Array Operations (Vectorized)**

```python
arr = np.array([1, 2, 3])

print (arr)
print(arr + 5)       # [6 7 8] - Adds 5 to each element
print(arr * 2)       # [2 4 6] - Multiplies each element by 2
print(arr ** 2)      # [1 4 9] - Squares each element
```

**📌 Indexing and Slicing**

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr)
print(arr[0])        # 10 - Accessing first element
print(arr[1:4])      # [20 30 40] - Slicing (Start from index 1 and go up to (but NOT including) index 4)
arr[2] = 99          # Modifying an element
print(arr)
```

# **📌 Basic Arithmetic Operations (Element-wise)**

```python
import numpy as np

# Creating two arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Addition
print(arr1 + arr2)  # Output: [5 7 9]

# Subtraction
print(arr1 - arr2)  # Output: [-3 -3 -3]

# Multiplication
print(arr1 * arr2)  # Output: [4 10 18]

# Division
print(arr1 / arr2)  # Output: [0.25 0.4 0.5]
```

**Scalar Operations (Applying a single value to all elements)**

```python
import numpy as np

# Creating two arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(arr1 + 10)  # Output: [11 12 13]  (Add 10 to each element)
print(arr2 * 2)   # Output: [ 8 10 12]  (Multiply each element by 2)
```

**Power and Square Root**

```python
import numpy as np

# Creating two arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(arr1 ** 2)         # Output: [1 4 9]  (Squares each element)
print(np.sqrt(arr2))      # Output: [2. 2.236 2.449]  (Square root of each element)
```

📌 Comparison Operations

```python
import numpy as np

# Creating two arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(arr1 > 2)           # Output: [False False True]
print(arr2 <= 5)          # Output: [ True  True False]
```

📌 Array-Wise Operations (Sum, Mean, etc.)

```python
import numpy as np

# Creating two arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(np.sum(arr1))        # Output: 6 (1 + 2 + 3)
print(np.mean(arr2))       # Output: 5.0
print(np.max(arr1))        # Output: 3
print(np.min(arr2))        # Output: 4
```

# 📊 Plotting & Visualization with Matplotlib

Matplotlib is a powerful Python library for creating static, interactive, and animated visualizations. It's built on top of NumPy and works seamlessly with it.

```python
#✅ Installing Matplotlib
!pip install matplotlib
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Sine Wave", color='blue')

# Adding Title and Labels
plt.title("Sine Wave", fontsize=14)
plt.xlabel("X Axis (0 to 10)", fontsize=12)
plt.ylabel("Y Axis (Amplitude)", fontsize=12)

# Customizing Ticks
plt.xticks(np.arange(0, 11, 1))    # X-axis ticks from 0 to 10 with step of 1
plt.yticks(np.arange(-1, 1.1, 0.5)) # Y-axis ticks from -1 to 1 with step of 0.5

# Adding a Grid for better readability
plt.grid(True)

# Show Plot
plt.show()

```

**🔍 Explanation:**

    plt.plot() - Plots y vs. x.

    plt.title(), plt.xlabel(), plt.ylabel() - Add labels and title.

    plt.show() - Displays the plot.

📌 Why Sine is on the Y-Axis:

Mathematical Function:

The sine function (np.sin()) is a common trigonometric function that oscillates between -1 and 1.

It represents periodic wave-like behavior which is perfect for visualization.

🔍 What This Code Does:

x = np.linspace(0, 10, 100)
y = np.sin(x)

**x is the Input:**

It’s a list of 100 numbers between 0 and 10.

Think of this as angles (in radians) going from 0 to 10.

**y is the Output:**

When we apply np.sin() to x, we get the sine values for each of those x values.

The result is a wave-like pattern because sin(x) repeats every 2π (approximately 6.28).

# **Using the csv Module (Built-in)**

The csv module provides basic functionality for reading and writing CSV fil

```python
#Reading a CSV File


import csv

with open("data.csv", mode="r",newline="", encoding="utf-8") as file:
    reader = csv.reader(file)  # Reads the file as a list of lists
    for row in reader:
        print(row)  # Each row is a list of values


#newline="": Prevents Python from adding extra newlines when reading the file.
#This is important when handling CSV files, as it ensures that each row is read correctly.
```

**Writing to a CSV File**


```python
#This is a list of lists, where:
#The first list ["Name", "Age", "City"] represents the CSV headers.
#Each subsequent list represents a row in the CSV file.

data = [
    ["Name", "Age", "City"],
    ["Alice", 25, "New York"],
    ["Bob", 30, "Los Angeles"]
]


#If "output.csv" does not already exist, Python will create a new file.
#If the file already exists, it will be overwritten because of the "w" (write) mode.

with open("output.csv", mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file) # creates a CSV writer object that allows you to write data to a CSV file
    writer.writerows(data)  # Writes multiple rows at once
```

```python
# Import the Flask class from the flask module
from flask import Flask

# Create an instance of the Flask class
app = Flask(__name__)

# Define a route (URL path) for the home page ('/')
@app.route('/')
def home():
    # This function is executed when someone accesses the home page
    return "Hello, Flask!"  # Returns a simple text response

# This ensures the app runs only when the script is executed directly
if __name__ == '__main__':
    # Start the Flask development server with debug mode enabled
    app.run(debug=True)

```

 **Flask Routing (URLs)**

Flask allows dynamic URL routes:

```python
# Import necessary libraries
from flask import Flask
from pyngrok import ngrok

# Initialize Flask app
app = Flask(__name__)

# Start an ngrok tunnel to expose Flask app to the internet
public_url = ngrok.connect(5000)  # Expose port 5000
print(f"🔗 ngrok Tunnel URL: {public_url}")  # Copy & open this link

# Define the root route
@app.route('/')
def home():
    return "Welcome! Try visiting /user/YourName"

# Define the dynamic route
@app.route('/user/<name>')
def greet_user(name):
    """Returns a personalized greeting with the given name."""
    return f"Hello, {name}!"

# Run the Flask app
if __name__ == '__main__':
    app.run(port=5000)
```

**Rendering HTML with Flask (Jinja2 Templating)**

Flask uses Jinja2 to generate dynamic web pages.

```python
import os

# Create templates folder
os.makedirs('templates', exist_ok=True)

# Create home.html template
with open("templates/home.html", "w") as f:
    f.write("""
    <!DOCTYPE html>
    <html lang="en">
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Flask with Jinja2</title>
    </head>
    <body>
        <h1>Welcome to Flask with Jinja2!</h1>
        <p>Hello, {{ name }}!</p>  <!-- This will be replaced with the 'name' variable -->
        <p>Visit your personalized page: <a href="/user/{{ name }}">Click here</a></p>
    </body>
    </html>
    """)
```

```python
from flask import Flask, render_template
from pyngrok import ngrok

# Initialize Flask app
app = Flask(__name__)

# Start ngrok tunnel to expose Flask app to the internet
public_url = ngrok.connect(5000)
print(f"🔗 ngrok Tunnel URL: {public_url}")  # Copy this URL to open the app in a browser

# Route to render a template with dynamic content
@app.route('/')
def home():
    name = "Alice"  # You can dynamically change this or make it user input
    return render_template('home.html', name=name)

# Route to handle user-specific URLs
@app.route('/user/<name>')
def greet_user(name):
    return f"Hello, {name}!"

# Run the Flask app on port 5000
if __name__ == '__main__':
    app.run(port=5000)
```

 # **🔹Key Components of Flask Microservices**

**1️⃣ API Endpoints**

Each microservice provides a RESTful API using Flask-RESTful libraries, allowing communication via HTTP methods like GET, POST, PUT, and DELETE. These APIs have specific endpoints for tasks like fetching user data or processing orders, ensuring independent and scalable microservices.

```python
# Import required libraries
from flask import Flask, jsonify
from pyngrok import ngrok
import threading

# Create a Flask application
app = Flask(__name__)

# Define a health check endpoint
@app.route('/health', methods=['GET'])
def health_check():
    return jsonify({"status": "healthy"}), 200

# Function to start the Flask app
def run_app():
    app.run(host="0.0.0.0", port=5000)  # Allow external access

# Start the Flask server in a separate thread
threading.Thread(target=run_app).start()

# Expose the local server using ngrok
public_url = ngrok.connect(5000).public_url
print(f"🌍 Public URL: {public_url}/health")  # Show correct endpoint URL
```

In Flask, jsonify is used to convert Python dictionaries (or lists) into JSON format for API responses

**2️⃣ Database Communication**

Each microservice manages its own database (e.g., PostgreSQL, MongoDB, MySQL), ensuring independence, scalability, and technology flexibility. Use SQLAlchemy for SQL databases or PyMongo for MongoDB.Microservices communicate via REST APIs, gRPC, or message queues to exchange data securely without direct database access.

**📌 User Service (Manages User Data)**

```python
from flask import Flask, jsonify
from flask_restful import Api, Resource
import threading

app_user = Flask(__name__)
api_user = Api(app_user)

# User Data (Mock Database)
users = {1: {"name": "Alice", "email": "alice@example.com"}}

class User(Resource):
    def get(self, user_id):
        user = users.get(user_id, None)
        if user:
            return jsonify({"id": user_id, "data": user})
        return jsonify({"message": "User not found"}), 404

api_user.add_resource(User, "/user/<int:user_id>")

def run_user_service():
    app_user.run(port=5001, debug=True, use_reloader=False)

# Start the User Service in a separate thread
threading.Thread(target=run_user_service).start()
```

**📌 Order Service** (Handles Orders, Calls User Service)

```python
from flask import Flask, jsonify
from flask_restful import Api, Resource
import requests

app_order = Flask(__name__)
api_order = Api(app_order)

orders = {101: {"user_id": 1, "item": "Laptop", "price": 1200}}

class Order(Resource):
    def get(self, order_id):
        order = orders.get(order_id, None)
        if order:
            # Call User Service to fetch user details
            user_response = requests.get(f"http://127.0.0.1:5001/user/{order['user_id']}")
            user_data = user_response.json() if user_response.status_code == 200 else {"message": "User not found"}
            return jsonify({"order_id": order_id, "order_details": order, "user": user_data})
        return jsonify({"message": "Order not found"}), 404

api_order.add_resource(Order, "/order/<int:order_id>")

def run_order_service():
    app_order.run(port=5002, debug=True, use_reloader=False)

# Start the Order Service in a separate thread
threading.Thread(target=run_order_service).start()
```

**📌 Payment Service** (Handles Payments, Calls Order Service)

```python
app_payment = Flask(__name__)
api_payment = Api(app_payment)

payments = {}

class Payment(Resource):
    def post(self, order_id):
        # Check if order exists
        order_response = requests.get(f"http://127.0.0.1:5002/order/{order_id}")
        if order_response.status_code == 200:
            payments[order_id] = "Paid"
            return jsonify({"message": f"Payment successful for Order {order_id}"})
        return jsonify({"message": "Order not found"}), 404

api_payment.add_resource(Payment, "/pay/<int:order_id>")

def run_payment_service():
    app_payment.run(port=5003, debug=True, use_reloader=False)

# Start the Payment Service in a separate thread
threading.Thread(target=run_payment_service).start()
```
